```python
sentence = "People walked down the street with lightness in their steps, smiles appearing effortlessly. A little boy laughed loudly as his balloon bobbed above him, and even a passing cyclist couldn’t help but grin at the sound."

words =sentence.split(" ") #split the sentence to words 

print(words)

#extracting the second word
print(words [1])

#extracting first 7 words
print(words [0:7])

#extracting last 2 words

print(words[-2:])

#print number of words

print("The number of words:", len(words))
```

The previous code printed the first word, the first seven words, the last two words, and the overall word count.
