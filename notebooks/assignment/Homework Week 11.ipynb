```python
import urllib.request 
text= urllib.request.urlopen("https://raw.githubusercontent.com/anettenuutmann/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/sentences.txt").read().decode("utf-8")

print(text[:300])
words = text.split()

print(words)

#extracting the second word
print(words [1])

#extracting first 7 words
print(words [0:7])

#extracting last 2 words

print(words[-2:])

#print number of words

print("The number of words:", len(words))
```

```python
from wordcloud import WordCloud

wc = WordCloud

import urllib.request #This one we use to work with URLs

text= urllib.request.urlopen("https://raw.githubusercontent.com/anettenuutmann/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/sentences.txt").read().decode("utf-8")

WordCloud(background_color="pink").generate_from_text(text).to_image()
```

The most often occurring words in the text are the biggest in the wordcloud. These words are: even, began, across. The smaller words, meaning they still appear frequently but less compared to the previous ones, are: wind, friends, soon and dog.
