```python
import pandas as pd
df= pd.read_csv("sports.csv")
df.head() # this is to show the first rows
```

```python
df.describe() # generate some statistics
```

Box plot to for each athlete to find the number of years of practice

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years of practice", legend=False)
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show 
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.ylabel("Age")
plt.title("Ages of Athletes")
plt.show ()
```

Age vs. years of Practice

plt.scatter (x,y) makes points for each pair of values
"s" controls the size of each bubble
s=df["Years of practice"]*30 means: if a student has more years of practice their bubble is larger. *30 is a scaling factor to make the bubble big

```python
plt.scatter(df["Age"], df["Years of practice"], s=df["Years of practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs years of practice")
plt.show
```

Average age per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Average age per sport")
plt.ylabel("Average age")
plt.show()
```
