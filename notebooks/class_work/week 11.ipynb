```python
sentence = "racecarlevelfield"

# extract first 3 letters

print("Print first 3 letters:", sentence[0:3]) #start: end

print(sentence[:3])

#extract letters from 5 to 9
print(sentence[5:9])

#extract letters from 5th to end 
print(sentence[5:])

#extract the last character

print(sentence[-1])

#Extract last 4 chracters
print(sentence[-4:])

#extract first letter of the substring starting from the 5
# sentence[5:] will extract a substring starting from 5
#0 this will take the fist character of that substring
print(sentence[5:][0])


#Extract first 5 letters of the substring starting from 5

print(sentence[5:][0:5])

print(sentence[:]) #extract entire text 


#extract every 2nd character 
print(sentence[::2])
```

```python
sentence = "Peter went to school and started singing"

words =sentence.split(" ") #split the sentence to words 

print(words)

#extract first word
print(words [0])

#extract first 3 words
print(words [0:3])

#extract last 3 words

print(words[-3:])


#print number of words

print("The number of words:", len(words))
```

```python
from wordcloud import WordCloud

from PIL import Image  #Import image class from PIL

import urllib.request #Library to work with url 

import numpy as np # in order to work with image masks

text =urllib.request.urlopen("https://drive.google.com/uc?export=download&id=1JjM4FNDMExTum90O4GqJMJcbiogm4lmA").read().decode("utf-8")


img =urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1A9NTBFy4H4AcfHrKY881dv6n_nM2JCGR",
"image1.png")

WordCloud(mask=np.array(Image.open("image1.png")),background_color="blue").generate_from_text(text).to_image()
```
