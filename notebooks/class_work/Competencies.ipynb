```python
import pandas as pd
df= pd.read_csv("Competencies.csv")
df.head() # this is to show the first rows
```

Who puts in the most effort

```python
import matplotlib.pyplot as plt
avg_hours=df.groupby("Student") ["Hours_Studied"].mean()
#Bar chart
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied")
#titles
plt.title("Who puts in the most effort?")
plt.ylabel("Hours studied")
plt.show()
```

Who performs the best 

```python
avg_hours=df.groupby("Student") ["Grade"].mean().sort_values(ascending=False)
#Bar chart
avg_hours.plot(kind="bar", x="Student", y="Grade", legend=False)
#titles
plt.title("Who performs best (By analysing average grades per student?")
plt.ylabel("Grade")
plt.show()
```

```python
#Average study hours per student
avg_grades=df.groupby("Student") ["Grade"].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 8))

#Bar chart of study hours per student 
avg_hours.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%')

#titles
plt.title("Who performs best (By analysing average grades per student?")
plt.ylabel("Grade")
plt.show()
```

Does more studying = better grades (Check exceptions!)

```python
#Does more studying= better grades

plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Does more studying=better grades? (Check exceptions!)")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
plt.show()
```

How consistent are study patterns and performance?

```python
#Calculate quartiles and extremes
hours_min= df["Hours_Studied"].min()
hours_q1= df["Hours_Studied"].quantile(0.25)
print("Minimum hours:", hours_min, "Q1:", hours_q1)

plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("How consistent are study patterns and performance?")
plt.ylabel("Value")
plt.show()

```
